```python
import pandas as pd
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
display(df.head())
```

```python
#Count the number of duplicate rows in the dataset.
dublicate_count = df.duplicated().sum()
print(f"Number of duplicate rows in the dataset: {dublicate_count}")
```

```python
#count of rows and columns in the dataset
rows, columns = df.shape
print(f"The dataset contains {rows} rows and {columns} columns.")
```

```python
#Display the first few duplicate rows to understand their structure
dublicates = df[df.duplicated()]  
display(dublicates.head())
```

```python
#Remove duplicate rows from the dataset using the drop_duplicates() function.
df = df.drop_duplicates()
```

```python
#verify the removal by counting the number of duplicate rows after removal .
dublicate_count_after_removal = df.duplicated().sum()
print(f"Number of duplicate rows after removal: {dublicate_count_after_removal}")
```

```python
#Identify missing values for all columns in the dataset.
how_many_missing_values = df.isnull().sum()
print("Missing values in each column:", how_many_missing_values)
```

```python
#Choose a column with significant missing values (e.g., EdLevel) and impute with the most frequent value.
# first, let's check the most frequent value in the 'EdLevel' column
most_frequent_value = df['EdLevel'].mode()
print("The most frequent value in the 'EdLevel' column is: ", most_frequent_value)

```

```python
# Impute missing values with the most frequent value
df['EdLevel'] = df['EdLevel'].fillna(most_frequent_value[0])
```

```python
# Check missing values after imputation
missing_values_after_imputation = df['EdLevel'].isnull().sum()

print("Missing values after imputation:", missing_values_after_imputation)
```
