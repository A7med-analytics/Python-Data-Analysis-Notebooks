```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
file_path = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/VYPrOu0Vs3I0hKLLjiPGrA/survey-data-with-duplicate.csv"
df = pd.read_csv(file_path)
# Display the first few rows
display(df.head())
```

```python
#Number of rows and columns in the dataframe
print("number of rows:", len(df))
print("number of columns:", len(df.columns))
```

```python
 #count the number of duplicate rows in the DataFrame
duplicate_rows = df.duplicated()    
print("Number of duplicate rows: ", duplicate_rows.sum())
```

```python
#Display the first few duplicate rows to understand their structure.
display(df[duplicate_rows].head(20))
```

```python
print (df.columns.tolist())
```

```python
#1. Identify duplicate response patterns based on selected columns such as MainBranch, Employment, and RemoteWork.
duplicate_patterns = df[
    df.duplicated(
        subset=['MainBranch', 'Employment', 'RemoteWork'],
        keep=False
    )
]

print("Number of duplicate response patterns:", len(duplicate_patterns))
```

```python
country_distribution = duplicate_patterns['Country'].value_counts()
print("Country distribution of duplicate response patterns:")
display(country_distribution)
```

```python
country_distribution.head(20).plot(kind='bar', figsize=(10, 6), color='skyblue')
plt.title('Country Distribution of Duplicate Response Patterns')    
```

```python
Employment_distribution = duplicate_patterns['Employment'].value_counts()
print("Employment distribution of duplicate response patterns:")    
display(Employment_distribution)
```

```python
Employment_distribution.head(10)
```

```python
Employment_distribution.head(10).plot(
    kind='bar',
    figsize=(12, 6),
    title='Respondents Distribution by Employment',
    xlabel='Employment Status',
    ylabel='Number of Respondents',
)

plt.tight_layout()
plt.show()
```
