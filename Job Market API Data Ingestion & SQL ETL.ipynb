```python
api_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"
```

```python
import requests
response = requests.get(api_url)
data = response.json()
```

```python
def get_number_of_jobs_T(technology):
    N_technology = 0
    for skill in data:
        if technology in skill['Key Skills']:
            N_technology += 1
    return N_technology
```

```python
get_number_of_jobs_T("Python")
```

```python
def get_number_of_jobs_T_location(LOC):
    N_LOC = 0
    for skill in data:
        if LOC in skill['Location']:
            N_LOC += 1
    return N_LOC
```

```python
get_number_of_jobs_T_location('Los Angeles')
```
