```python
import pandas as pd
import requests
```

```python
file_path = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/VYPrOu0Vs3I0hKLLjiPGrA/survey-data-with-duplicate.csv"
```

```python
db=pd.read_csv(file_path)
```

```python
#display the first 5 rows of the dataframe
db.head(5)
```

```python
#number of rows and columns in the dataframe
print("number of rows:", len(db))
print("number of columns:", len(db.columns))
```

```python
#Print the datatype of all columns
print("Data types of all columns:", db.dtypes)
```

```python
# find Unique values in the 'Age' column
db['Age'].unique()
```

```python
age_mapping = {
    "Under 18 years old": 17,
    "18-24 years old": 21,
    "25-34 years old": 29,
    "35-44 years old": 39,
    "45-54 years old": 49,
    "55-64 years old": 59,
    "65 years or older": 70 }  
```

```python
# Create a new numeric Age column
db['Age_numeric'] = db['Age'].map(age_mapping)  
```

```python
db['Age_numeric'].unique()
```

```python
# find mean of the Age_numeric column
mean_age = db['Age_numeric'].mean()
print("Mean of Age_numeric column:", mean_age) 
```

```python
# find the number of unique countries in the 'Country' column
num_unique_countries = db['Country'].nunique()
print("Number of unique countries in the 'Country' column:", num_unique_countries)
```
